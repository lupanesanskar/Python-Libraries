# **Numpy Basic Functions**

### What is numpy?

NumPy is a basic and important Python library for scientific computing. It provides a multidimensional array that allows us to store and work with large amounts of data easily. It also provides many functions for mathematical calculations, logical operations, changing the shape of data, sorting, selecting data, statistics, linear algebra, random numbers, and more.

The main object in NumPy is called ndarray. It is used to store one-dimensional, two-dimensional, or multi-dimensional arrays. The elements in a NumPy array generally have the same data type

### Numpy Arrays Vs Python Sequences

- NumPy arrays have a fixed size when they are created, while Python lists can grow or shrink easily. If the size of a NumPy array needs to be changed, a new array is generally created.

- The elements of a NumPy array are usually of the same data type, so they take the same amount of memory. Python lists can contain different types of data.

- NumPy arrays are designed for fast mathematical and numerical operations on large amounts of data. These operations are usually faster and require less code compared to using normal Python lists.

- NumPy is also widely used by other scientific, Data Science, and Machine Learning libraries. These libraries often convert Python lists or other sequences into NumPy arrays before processing the data and may also return NumPy arrays as the result.

## 2) Slicing

```python
m1
```

```python
print(m1[2:5:2])
```

```python
m2
```

```python
print(m2[0,:])
print(m2[1,::3])
print(m2[::2,0])
print(m2[1:,2:])
print(m2[:,1:])
print(m2[::2,::3])
print(m2[::2,::2])
```

```python
m3=np.arange(27).reshape(3,3,3)
m3
```

```python
print(m3[1])  # For 1st matrix 
print(m3[1,:,1])  # for 2nd matrix 2 column
print(m3[2,1:,1:]) # for 3rd matrix bottom right 2x2 matrix 
print(m3[::2,0,::2]) 
```
