## np.sort

```python
import numpy as np
np.random.seed(7)               # seed -> same random numbers every run, so the output always matches
a=np.random.randint(1,100,15)   # 15 random integers from 1 to 99 (100 is excluded)
a
```

```python
# For appending in numpy Array
np.append(a,100)   # adds 100 at the end and returns a NEW array
                   # 'a' itself is NOT changed (check by running 'a' again)
```

```python
np.sort(a) # numpy sorting gives you numpy array and in ascending order
           # original 'a' stays the same, sort returns a sorted copy
```

```python
np.sort(a)[::-1] # [::-1] reverses the sorted array -> descending order
```

```python
b=np.random.randint(1,100,24).reshape(6,4)  # 24 numbers arranged as 6 rows x 4 columns
print(b)                                     # original 2D array
np.sort(b,axis=0) # axis = 0 -> every COLUMN is sorted top to bottom (ascending)
                  # axis = 1 -> every ROW is sorted left to right
```

```python
# For appending in 2D array
np.append(b,np.ones((b.shape[0],1)),axis=1)  # axis=1 -> add a new COLUMN at the right side
# np.ones((6,1)) makes a column of ones; the double bracket is required
# because np.ones takes ONE argument (the shape) and that shape is a tuple
# b.shape[0] = no. of rows
# b.shape[1] = no. of columns
# output is float because np.ones gives floats (1.), so all numbers become float
```

```python
np.append(b,np.ones((1,4)),axis=0) # axis=0 -> add a new ROW at the bottom
# the new row must have the same no. of columns as b (4), that's why shape is (1,4)
# if not understand see np.ones function in Fundamentals file
```

## np.concatinate

```python
m=np.arange(2,8).reshape(2,3)     # numbers 2 to 7 in shape 2 x 3
n=np.arange(20,26).reshape(2,3)   # numbers 20 to 25 in shape 2 x 3
print(m)
print(n)
```

```python
np.concatenate((m,n),axis=1) # axis = 1 -> joins side by side (columns increase: 3+3 = 6 columns)
# note: arrays are passed inside ONE tuple (m,n)
```

```python
np.concatenate((m,n),axis=0) # axis = 0 -> joins one below the other (rows increase: 2+2 = 4 rows)
```

## np.expand_dims
It coverts the 1D  array to 2D array similarly for 3D to 4D means Its Add only one dimension in available dimension

```python
a=np.array([3,12,7,9,1,450,7,28,12,7,3,9])
a.shape   # (12,) -> 1D array with 12 elements
```

```python
np.expand_dims(a,axis=1)   # axis=1 -> adds a new axis at the end: shape (12,) becomes (12,1)
                           # every element now sits in its own row (column vector)
```

```python
np.expand_dims(a,axis=0)   # axis=0 -> adds a new axis in front: shape (12,) becomes (1,12)
                           # whole array is now ONE row (row vector)
```

## np.where

```python
b=np.random.randint(1,100,15)   # 15 random numbers from 1 to 99
b
```

```python
np.where(b>50)  # It gives index numbers (positions) of the values which are greater than 50
                # output is a tuple, the array inside it contains the indexes
```

```python
# replace all values > 50 with zero
# np.where(condition,true,false)
np.where(b>50,0,b)   # if b>50 is True -> put 0, else -> keep the original value of b
                     # returns a new array, b is unchanged
```

## np.cumsum (cumulative sum)
like fibonacchi series

```python
a=np.array([8,91,23,4,17,56,38,12,70,29,45,6])
np.cumsum(a)   # cumulative sum -> each value = sum of all values up to that position
               # e.g. 8, 8+91=99, 99+23=122 ...
```

```python
b=np.array([[9,31,14],[22,5,47],[63,18,7],[12,40,26]])
b
```

```python
b=np.array([[9,31,14],[22,5,47],[63,18,7],[12,40,26]])
np.cumsum(b,axis=1) # Row wise -> running total goes left to right inside each row
                    # axis=0 would give column wise running total
```

```python
np.cumprod(a) # similarly for product -> each value = product of all values up to that position
              # grows very fast, so numbers become big
```

## np.corrcoef 
It gives relation between two or more  tham two arrays in the for of number.


```python
salary=np.array([18000,42000,30000,75000,55000])
experience=np.array([1,3,2,7,5])
np.corrcoef(salary,experience)
# gives correlation matrix: value is between -1 and +1
# close to +1 -> both increase together, close to -1 -> one increases other decreases, 0 -> no relation
# diagonal is always 1 (an array is fully related to itself)
```
