###`Q-1:` You are given a 4x3 matrix. Perform the below tasks on it.

    i. Sort every row of the matrix. np.sort()
    ii. Sort the rows of the matrix based on the values of the 3rd column (largest first).
    iii. Sort the rows of the matrix based on the sum of each row (smallest first).
    iv. Sort all the elements of the matrix and keep the same shape.
    v. Sort the rows based on the 1st column, and if two values of the 1st column are same then use the 2nd column.

```
arr = [[45 12 78]
       [33 90 21]
       [45  5 54]
       [12 88 40]]
```
```
i. np.sort
[[12 45 78]
 [21 33 90]
 [ 5 45 54]
 [12 40 88]]

ii. based on 3rd column - largest first
[[45 12 78]
 [45  5 54]
 [12 88 40]
 [33 90 21]]

iii. based on row sum - smallest first
[[45  5 54]
 [45 12 78]
 [12 88 40]
 [33 90 21]]

iv. based on elements value
[[ 5 12 12]
 [21 33 40]
 [45 45 54]
 [78 88 90]]

v. based on 1st column then 2nd column
[[12 88 40]
 [33 90 21]
 [45  5 54]
 [45 12 78]]
```
Hint: for the last one read about `np.lexsort()`. Be careful, it uses the **last** key as the main key.

````python
#Code here
arr = np.array([[45, 12, 78],
                [33, 90, 21],
                [45, 5, 54],
                [12, 88, 40]])

# i. np.sort sorts every row separately
print(np.sort(arr))
````

````python
# ii. argsort gives the row order, [::-1] reverses it so the largest comes first
print(arr[arr[:, 2].argsort()[::-1]])
````

````python
# iii. sum of every row, then sort rows by that
print(arr[arr.sum(axis=1).argsort()])
````

````python
# iv. axis=None makes it flat before sorting, so we need reshape after that
print(np.sort(arr, axis=None).reshape(arr.shape))
````

````python
# v. lexsort takes the keys as (second key, main key)
print(arr[np.lexsort((arr[:, 1], arr[:, 0]))])
````

###`Q-2:` There is an array of marks of 4 students in 3 subjects. Further you are asked to perform below tasks.

    i. Add the marks of an extra subject for every student in the same array.
    ii. Add two new students marks in respective 4 subjects (one subject added in above task).
    iii. Add an extra column with the sum of all subjects (4 subjects) marks.
    iv. Sort the array (non-ascending order) on the total marks column (the one added in above task). Show the top 3 rows.
    v. Find the row numbers of the students who scored more than 60 in every subject (do not use the total column here).

Note: Change dimension of arrays during concatenation or appending if required.

Given Array-
```
marks = [[34, 56, 78],
         [45, 89, 23],
         [90, 67, 81],
         [12, 44, 38]]

extra_subject = [65, 70, 55, 48]
# Two extra students record-
rec1 = [88, 72, 91, 67]
rec2 = [40, 35, 58, 62]
```
Result of task iv-
```
[[ 88  72  91  67 318]
 [ 90  67  81  55 293]
 [ 34  56  78  65 233]]
```

````python
#Code here
marks = np.array([[34, 56, 78],
                  [45, 89, 23],
                  [90, 67, 81],
                  [12, 44, 38]])

extra_subject = np.array([65, 70, 55, 48])
rec1 = np.array([88, 72, 91, 67])
rec2 = np.array([40, 35, 58, 62])

# i. extra_subject is 1D, so make it a column (4,1) before adding it
marks = np.append(marks, extra_subject.reshape(-1, 1), axis=1)
marks
````

````python
# ii. vstack adds the new rows at the bottom
marks = np.vstack((marks, rec1, rec2))
marks
````

````python
# iii. sum of each row, made as a column and added at the end
marks = np.append(marks, marks.sum(axis=1).reshape(-1, 1), axis=1)
marks
````

````python
# iv. sort using the last column, largest first
sorted_marks = marks[marks[:, -1].argsort()[::-1]]
sorted_marks[:3]
````

````python
# v. [:, :-1] leaves the total column, all(axis=1) is True if the whole row is True
np.where((marks[:, :-1] > 60).all(axis=1))[0]
````

###`Q-3:` Find unique arrays from a 2D array row wise and column wise. Also,

### `Q-5:` Given a NumPy array and a number X, find the below things.

    i. Row numbers which have at least one element larger than X.
    ii. Row numbers in which all the elements are larger than X.
    iii. How many elements are larger than X in every row.

```
arr = [[ 5, 12,  7,  9],
       [15, 18, 20, 11],
       [ 2, -4,  8,  6],
       [10, 13, 14, 16]]

X = 9
```
Result-
```
i.   [0 1 3]
ii.  [1 3]
iii. [1 4 0 4]
```

````python
#Code here
arr = np.array([[5, 12, 7, 9],
                [15, 18, 20, 11],
                [2, -4, 8, 6],
                [10, 13, 14, 16]])
X = 9

# arr > X is a True/False matrix, any(axis=1) checks every row
print(np.where((arr > X).any(axis=1))[0])
````

````python
# all(axis=1) is True only if the whole row is True
print(np.where((arr > X).all(axis=1))[0])
````

````python
# True is counted as 1, so the sum of every row is the count
print((arr > X).sum(axis=1))
````

    ii. Flatten the 2D array in row order and also in column order.

```
# These arrays are given.
a1 = np.arange(4)
```

### `Q-7:` You are given an array. You have to find the minimum and maximum value of the array and remove **all** the places where these two values are present. Keep the array 2D (shape should be `(1, n)`) and also print how many elements are removed.

```python
import numpy as np

np.random.seed(10)
arr = np.random.randint(10, 40, 60).reshape((1, 60))
```

````python
#Code here
np.random.seed(10)
arr = np.random.randint(10, 40, 60).reshape((1, 60))

print(arr.min(), arr.max())

# keep only those which are not min and not max
mask = (arr != arr.min()) & (arr != arr.max())

# boolean mask gives a 1D array, reshape(1, -1) makes it 2D again
new_arr = arr[mask].reshape(1, -1)
print(new_arr.shape)
print("removed:", arr.size - new_arr.size)
````

````python
# OR : np.delete does not change the original array, so we have to save the result
index = np.where(~mask)[1]
np.delete(arr, index, axis=1).shape
````

### `Q-8:` You are given an array of 20 numbers between 100 and 700. Limit the elements of this array between 200 and 500, so replace the smaller values with 200 and the bigger values with 500. Then,
- print how many elements were changed by this.
- sort the array in descending order and perform the cumulative sum of that array.
- check that the last value of the cumulative sum is equal to the sum of the array.

```
np.random.seed(5)
arr = np.random.randint(100, 701, 20)
```

````python
#Code here
np.random.seed(5)
arr = np.random.randint(100, 701, 20)

clipped = np.clip(arr, 200, 500)
print(clipped)

# compare old and new, True means that place was changed
print("changed:", (arr != clipped).sum())
````

````python
sorted_arr = np.sort(clipped)[::-1]
csum = np.cumsum(sorted_arr)
print(csum)
print(csum[-1] == clipped.sum())
````

### `Q-9:` You are given an array of 40 random numbers between 0 and 1. First round off the elements up to 3 decimal places and then compare that
- 0th percentile == minimum value of the array
- 100th percentile == maximum value of the array
- 50th percentile == median of the array
- also find the difference between 51st percentile and 50th percentile values
- also find the IQR (75th percentile - 25th percentile) of the array

```
np.random.seed(1)
arr = np.random.rand(40)
```

````python
#Code here
np.random.seed(1)
arr = np.random.rand(40)

# round the elements first, as the question says
arr = np.round(arr, 3)

print(np.percentile(arr, 0) == arr.min())
print(np.percentile(arr, 100) == arr.max())
print(np.percentile(arr, 50) == np.median(arr))
````

````python
print(np.round(np.percentile(arr, 51) - np.percentile(arr, 50), 3))

# IQR : middle 50% data spread
print(np.round(np.percentile(arr, 75) - np.percentile(arr, 25), 3))
````

###`Q-11:` Split the array in equal parts and find the sum of every part.

    i. Split the array of 12 numbers in 3 equal parts.
    ii. Now try to split an array of 10 numbers in 3 parts. What happens? Find another way to do it and print the size of every part.

```
arr1 = np.arange(1, 13)
arr2 = np.arange(1, 11)
```

````python
#Code here
arr1 = np.arange(1, 13)
parts = np.split(arr1, 3)
print(parts)
print([int(p.sum()) for p in parts])
````

````python
arr2 = np.arange(1, 11)

# np.split(arr2, 3) will give error, as 10 can not be divided equally in 3 parts
# array_split does not need equal parts, the first parts get the extra elements
parts = np.array_split(arr2, 3)
print(parts)
print([p.size for p in parts])
````
